```python
# best.pt 업로드 (파일 선택 창이 뜨면 best.pt 선택)
from google.colab import files

uploaded = files.upload()
MODEL_PATH = next(iter(uploaded))  # 업로드한 파일명 그대로 사용
print("model:", MODEL_PATH)
```

```python
from ultralytics import YOLO

model = YOLO(MODEL_PATH)
print(model.names)
```

```python
# 분석할 영상(또는 이미지) 업로드 (여러 개 선택 가능)
uploaded_videos = files.upload()
SOURCE = next(iter(uploaded_videos))  # 여러 개 올렸으면 list(uploaded_videos) 중 원하는 파일명 사용
print("source:", SOURCE)
```

```python
results = model.predict(
    source=SOURCE,
    conf=0.35,
    imgsz=960,
    save=True,       # runs/predict 에 결과(박스 그려진 영상/이미지) 저장
    project="runs",
    name="predict"
)
```

```python
# 이미지 결과면 미리보기 (영상은 다음 셀에서 zip으로 받아서 로컬로 재생해서 확인)
import glob
from IPython.display import Image, display

for p in sorted(glob.glob("runs/predict*/*.jpg"))[:5]:
    display(Image(filename=p))
```

```python
# 결과 폴더를 zip으로 묶어서 로컬로 다운로드
import shutil

shutil.make_archive("predict_results", "zip", "runs/predict")
files.download("predict_results.zip")
```
